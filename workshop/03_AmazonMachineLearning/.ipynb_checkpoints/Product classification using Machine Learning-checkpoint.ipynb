## Product classification using Multinomial Logistic Regression

Author: Samir Araujo (arsamir@amazon.com)

Our customer has a portal for selling products online. Each product must be classified in a given category and subcategory. The customer also has a trusted dataset, which contains a list of products and their respectives categories and subcategories. 

Every day a new product is added to the marketplace area, by external sellers. These products eventually are added in wrong categories and/or subcategories.

The idea of this task is to use the trusted dataset and to train a Machine Learning solution, capable of predicting the right classification for each new product. 

So, lets train two Multinomial Logistic Regression models for:
* Given the product title, predicts the category
* Given the product title and the category, predicts the subcategory

Once Logistic Regression is a numerical model, we must to transform the product title into numbers.
Lets use Term Frequency/Inverse Document Frequency (https://en.wikipedia.org/wiki/Tf%E2%80%93idf) to do that.

In order to obtain a homogeneous frequency table, we will first:
* Remove accents
* Remove stopwords (https://en.wikipedia.org/wiki/Stop_words)
* Tokenize title
* Compute ngram(2,2) for each product
* Create a vocabulary (list) containing each token computed in the previous step
* Compute the sparse matrix with the Term Frequency of each entry from the vocabulary for each product

The sparse matrix from step 6 will be the input of both regression models.

So, lets do it! \m/

```python
%matplotlib inline

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

#from patsy import dmatrices
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn import preprocessing
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split

plt.style.use('ggplot')
```

Lets load a sample of our dataset to work on it

```python
data = pd.read_csv('dataset/data.csv', sep=';', encoding='utf-8')
print( "Dataset size [before cleaning]: {}".format( len(data) ))
data = data[ (data.product_name.apply(type) == str) & 
             (data.main_category.apply(type) == str) & 
             (data.sub_category.apply(type) == str) 
           ]
print( "Dataset size [after cleaning]: {}".format( len(data) ))
data.head()
```

```python
main_data = data[['product_name', 'main_category', 'sub_category']]
main_data.groupby(['main_category', 'sub_category']).describe()
```

```python
main_data_count = main_data[['product_name', 'main_category']]
main_data_count = pd.DataFrame({'count' : main_data_count.groupby( [ "main_category"] ).size()}).reset_index()
main_data_count = main_data_count[ [ 'count', 'main_category' ] ]
main_data_count = main_data_count.set_index('main_category')
main_data_count
```

Now, lets load the stopwords table and remove accents from it. Our dataset is in portuguese, that's why we need to do that.

```python
# tranlating table for removing accents
accents = "".maketrans("áàãâéêíóôõúüçÁÀÃÂÉÊÍÓÔÕÚÜÇ", "aaaaeeiooouucAAAAEEIOOOUUC")

# loading stopwords without accents
file = open("stopwords.txt", "r")
stopwords = list(map(lambda x:x.strip().translate(accents),file.readlines()))
file.close()
```

Ok. Now we will create a word tokenizer that will be applied to the dataset

```python
# this tokenizer will tokenize the text, remove stop words and compute bigrams (ngram(2))
word_vectorizer = TfidfVectorizer(ngram_range=(1,2), analyzer='word', stop_words=stopwords, token_pattern='[a-zA-Z]+')
tokenizer = word_vectorizer.build_tokenizer()

def remove_stop_words(text):
    return " ".join( list(filter( lambda x: x not in stopwords, tokenizer(text) )) )
```

Using the tokenizer defined above, lets create new columns in our dataframe with the transformed fields

```python
data['product_name_tokens'] = list(map(lambda x: remove_stop_words( x.lower().translate(accents) ), data['product_name']))
data['main_category_tokens'] = list(map(lambda x: remove_stop_words( x.lower().translate(accents) ), data['main_category']))
data['subcategory_tokens'] = list(map(lambda x: remove_stop_words( x.lower().translate(accents) ), data['sub_category']))
data['subcategory_tokens'] = data['subcategory_tokens'].apply(lambda x: x.split(' ')[0] )
```

```python
output = data[['product_name_tokens', 'main_category_tokens', 'subcategory_tokens']]
output.columns = ['product_name', 'category', 'sub_category']

output.to_csv( 'data_tratado.csv', sep=',', encoding='utf-8', index=False)
```

Done. Now, using this new column, lets create the vocabulary list and the Sparse Matrix.
Please, notice that the vocabulary is the key for extracting the Term Frequency from the tokenized title column.

Lets split our dataset into two distinct datasets. One for training and another for evaluation

```python
train_data, test_data = train_test_split(data, test_size=0.3)
print( "Train dataset size: {}".format( len(train_data) ))
print( "Test dataset size: {}".format( len(test_data) ))
```

```python
word_vectorizer = TfidfVectorizer(ngram_range=(1,2), analyzer='word', stop_words=stopwords, token_pattern='[a-zA-Z]+')
sparse_matrix = word_vectorizer.fit_transform(data['product_name_tokens'])

dtm_train = word_vectorizer.transform( train_data['product_name_tokens'] ).toarray()
dtm_test = word_vectorizer.transform( test_data['product_name_tokens'] ).toarray()

vocab = np.array(word_vectorizer.get_feature_names())

print( "Vocabulary size: {}".format(len(vocab) ))
print( "Vocabulary sample: {}".format(vocab[100:150] ))
```

Ok. Now we need to convert the categories and subcategories from text to an enumeration.

```python
category_label_encoder = preprocessing.LabelEncoder()
category_label_encoder.fit( train_data['main_category_tokens'] )
categories_train = category_label_encoder.transform(train_data['main_category_tokens']) 

subcategory_label_encoder = preprocessing.LabelEncoder()
subcategory_label_encoder.fit( train_data['subcategory_tokens'])
sub_categories_train = subcategory_label_encoder.transform(train_data['subcategory_tokens'])
```

```python
category_label_encoder = preprocessing.LabelEncoder()
category_label_encoder.fit( test_data['main_category_tokens'] )
categories_test = category_label_encoder.transform(test_data['main_category_tokens']) 

subcategory_label_encoder = preprocessing.LabelEncoder()
subcategory_label_encoder.fit( test_data['subcategory_tokens'])
sub_categories_test = subcategory_label_encoder.transform(test_data['subcategory_tokens'])
```

Finally, before we train the final model on Amazon Machine Learning, let's train a model here.

Please, note that this is just a sample of the dataset. It will not give you the same score of a complete dataset. So, keep in mind that it is (again) just an idea of what will happen when you train a model on Amazon Machine Learning.

```python
# instantiate a logistic regression model, and fit with X and y
categories_model = LogisticRegression(solver='lbfgs', class_weight='balanced', penalty='l2', verbose=1, C=0.0001, random_state=42, multi_class='multinomial')
categories_model = categories_model.fit(dtm_train, categories_train)

# check the accuracy on the training set
categories_model.score(dtm_test, categories_test)
```

## Predicting
